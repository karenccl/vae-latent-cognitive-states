# 01 Explore the simulator
Sanity checks of the hidden-state simulator before any modelling.

```python
import sys; sys.path.insert(0, '../src')
import numpy as np, pandas as pd
from simulate import *
from data import build_dataset, load_config
cfg = load_config(); d = build_dataset(cfg)
print({k: d[k]['X'].shape for k in ('train','val','test')})
```

```python
ds = simulate_dataset(200, 300, seed=2026)
s = np.concatenate([x['states'] for x in ds]); rt = np.concatenate([x['log_rt'] for x in ds]); c = np.concatenate([x['correct'] for x in ds])
print('empirical state props', np.bincount(s)/len(s), 'stationary', stationary_distribution().round(3))
print(pd.DataFrame({'state': s, 'correct': c, 'log_rt': rt}).groupby('state').agg(['mean','std']).round(3))
```

```python
print('window label counts (train):', np.bincount(d['train']['y'].ravel(), minlength=3))
print('Windows are mixtures of states: share of windows where the majority state covers <10 trials:')
W = d['train']['states'].reshape(-1, 10)
print(np.mean([np.bincount(w, minlength=3).max() < 10 for w in W]).round(3))
```
